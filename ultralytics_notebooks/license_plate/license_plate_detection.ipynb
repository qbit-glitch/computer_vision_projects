```python
import cv2
from ultralytics import YOLO
import os

model = YOLO("/Users/qbit-glitch/Desktop/coding-projects/computer_vision_projects/runs/detect/train17/weights/best.pt")

IMAGE_FILE_PATH = "testing_images/sample_img1_crop3.jpeg"
results = model.predict(source=IMAGE_FILE_PATH, conf=0.25)

save_dir = "predicted_bounding_boxes"

# Load original Image
img = cv2.imread(IMAGE_FILE_PATH)

for r_idx, r in enumerate(results):       # for each image (useful if batch)
    for b_idx, box in enumerate(r.boxes): # for each detected box
        x1, y1, x2, y2 = box.xyxy[0].cpu().numpy().astype(int)
        cropped = img[y1:y2, x1:x2]

        filename = f"crop_img{r_idx}_box{b_idx}.jpg"

        cv2.imwrite(os.path.join(save_dir, filename), cropped)        
```
