```python

from ultralytics import YOLO
import torch

device = 'mps' if torch.backends.mps.is_available() else 'cpu'

model = YOLO('yolo11s.pt')

model.train(data='mnit_parking_dataset.yaml', epochs=3, imgsz=640, device=device)
```
