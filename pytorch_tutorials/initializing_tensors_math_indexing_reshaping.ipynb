#### How to initialize and convert tensors to other types (int, float, double)

```python
tensor = torch.arange(4)
print(tensor.bool())    # convert values to bool
print(tensor.short())   # convert values to int16
print(tensor.long())    # convert values to float16
print(tensor.half())    # float16
print(tensor.float())   # float32
print(tensor.double())  # float64
```

#### Array to Tensor conversion and vice versa


```python
import numpy as np

np_array = np.zeros((5,7))
print(np_array)

tensor = torch.from_numpy(np_array)
print(tensor)

np_array_back = tensor.numpy()
print(np_array_back)
```

## Tensor Math and Comparison Operations


```python
x = torch.tensor([1,2,3])
y = torch.tensor([9,8,7])
```

#### Addition

```python
z1 = torch.empty(3)
torch.add(x, y, out=z1)
print(z1)
print(z1.dtype)
```

```python
z2 = torch.add(x,y)
print(z2)
print(z2.dtype)

z3 = x + y
print(z3)
```

#### Subtraction

```python
z = x - y
print(z)
```

#### Division

```python
z = torch.true_divide(x,y)
print(z)
```

#### Inplace Operations

```python
t = torch.ones(3)
t.add_(x)
t += x
print(t)
```

#### Exponentiation

```python
z = x.pow(2)
print(z)

z = y ** 2
print(z)
```

#### Simple Comparison

```python
z = x > 2   # Assign bool value for every element based on the condition x>2
print(z)

z = y > 2
print(z)
```

#### Matrix Multiplication

```python
x1 = torch.rand((2,5))
x2 = torch.rand((5,3))
print(x1)
print(x2)

x3 = torch.mm(x1, x2)
print(x3)

x4 = x1.mm(x2)
print(x4)
```

#### Matrix Exponentiation

```python
matrix_exp = torch.rand(5,5)
print(matrix_exp)
print(matrix_exp.matrix_power(3))
```

#### Element Wise Multiplication

```python
z = x * y
print(z)
```

#### Dot Product

```python
z = torch.dot(x,y)
print(z)
```

#### Other useful tensor operations

```python
x = torch.tensor([[10,2,3,4], [8,4,0,9]], dtype=torch.float32, requires_grad=True, device=device)

# dim=0 will aggregate each columns
# dim=1 will aggregate each row
sum_x = torch.sum(x, dim=0)
print(sum_x)



values, indices = torch.max(x, dim=0)
print(values,"\n", indices)
```

```python
values, indices = torch.min(x, dim=0)
print(values, indices)
```

```python
abs_x = torch.abs(x)
print(abs_x)
```

```python
z = torch.argmax(x, dim=0)
print(z)

z = torch.argmin(x, dim=1)
print(z)
```

#### Notes:

- The torch.argmax() function in PyTorch returns the indices of the maximum values in an input tensor along a specified dimension or across all dimensions if no dimension is specified.
 If the tensor has multiple maximum values, it returns the index of the first occurrence.
 The function is not differentiable and should be avoided in gradient computations.

- The torch.argmin() function in PyTorch returns the indices of the minimum value(s) of the input tensor, either across the entire flattened tensor or along a specified dimension.
 The output tensor's index type is torch.int64.
 If multiple minimum values exist, it returns the index of the first occurrence

```python
mean_x = torch.mean(x.float(), dim=0)
print(mean_x)
```

```python
x = torch.tensor([1,2,3])
y = torch.tensor([9,2,7])

z = torch.eq(x, y)
print(z)
```

```python
sorted_y, indices = torch.sort(y, dim=0, descending=True)
print(sorted_y)
```

```python
x = torch.tensor([-1, -4, -9,8,9,2, 0, 1,2,3])
z = torch.clamp(x, min=0, max=3)
print(z)
```

## Tensor indexing

```python
batch_size = 10
features = 25

x = torch.rand((batch_size, features))
print(x)
```

```python
print(x[0].shape)   # x[0,:]

print(x[:,0].shape)

print(x[2, 0:10])
```

```python
x[0,0] = 100
print(x[:,0])
```

#### Fancy Indexing

```python
x = torch.rand(10)
print(x)
indices = [2,5,8]
print(x[indices])
```

```python
x = torch.rand((3,5))
rows = torch.tensor([1,0])
cols = torch.tensor([4,0])
print(x)
print(x[rows, cols])
```

#### More advanced Indexing

```python
x = torch.arange(10)

print(x[(x < 3) | (x > 8)])
print(x[x.remainder(2) == 0])
```

#### Useful Operations

```python
print(torch.where(x > 5, x, x * 2))
print(torch.tensor([0,0,1,2,2,3,4]).unique())
print(x.ndimension())
print(x.numel())
```

## Tensor Reshaping

```python
x = torch.arange(9)

x_3x3 = x.view(3,3)
print(x_3x3)
x_3x3 = x.reshape(3,3)
print(x_3x3)
```

```python
y = x_3x3.t()   
print(y)
```

```python
print(y.contiguous().view(9))
```

```python
x1 = torch.rand((2,5))
x2 = torch.rand((2,5))
print(torch.cat((x1,x2), dim=0))
print(torch.cat((x1,x2), dim=1))
```

```python
z = x1.view(-1)
print(z.shape)
```

```python
batch = 5
x = torch.rand((batch, 2, 5))
print(x)

z = x.view(batch, -1)
print(z)
```

```python
z = x.permute(0,2,1)
print(z)
print(z.shape)
```

```python
x = torch.arange(10)
print(x.unsqueeze(0))
print(x.unsqueeze(1))
```

```python
x = torch.arange(10).unsqueeze(0).unsqueeze(1)
print(x.shape)

z = x.squeeze(1)
print(z.shape)
```
